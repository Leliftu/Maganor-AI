```python
from sklearn.feature_extraction.text import TfidfVectorizer

sample_comments = [
    "you are a kind person",
    "you are an insulting person",
    "you are not a kind person"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(sample_comments)

# Convert the result into a readable table
import pandas as pd

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(tfidf_df.round(3))
```

```python
plt.figure(figsize=(10, 5))

sns.heatmap(
    tfidf_df,
    annot=True,
    cmap="Purples",
    fmt=".2f"
)

plt.title("TF-IDF Word Importance Across Comments")
plt.xlabel("Words")
plt.ylabel("Comments")

plt.tight_layout()
plt.show()
```

```python
# Input: comments
X = df["clean_text"]

# Target: toxicity labels
y = df["toxic"]

print("Number of comments:", len(X))
print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentages:")
print(y.value_counts(normalize=True) * 100)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training comments:", len(X_train))
print("Testing comments:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

# Learn vocabulary and transform training comments
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform testing comments using the same vocabulary
X_test_tfidf = tfidf.transform(X_test)

print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)

print("Number of learned features:", len(tfidf.get_feature_names_out()))
```

```python
from sklearn.linear_model import LogisticRegression

# Initialize the classifier
model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

# Train the model
model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
# Predict labels for unseen testing comments
y_pred = model.predict(X_test_tfidf)

print("First 20 actual labels:")
print(y_test.iloc[:20].values)

print("\nFirst 20 predicted labels:")
print(y_pred[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("MODEL PERFORMANCE")
print("-------------------------")
print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1-score:", round(f1, 4))

print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd

# Create a table of actual and predicted results
results = pd.DataFrame({
    "comment": X_test.values,
    "actual": y_test.values,
    "predicted": y_pred
}, index=X_test.index)

# Identify mistakes
false_positives = results[
    (results["actual"] == 0) &
    (results["predicted"] == 1)
]

false_negatives = results[
    (results["actual"] == 1) &
    (results["predicted"] == 0)
]

print("False positives:", len(false_positives))
print("False negatives:", len(false_negatives))
```

```python
# Get the probability of toxicity for each test comment
toxic_probabilities = model.predict_proba(X_test_tfidf)[:, 1]

# Add probabilities to our results table
results["toxic_probability"] = toxic_probabilities

# Inspect false positives with highest confidence
print("FALSE POSITIVES — HIGH CONFIDENCE")

print(
    results[results["actual"] == 0]
    .sort_values("toxic_probability", ascending=False)
    .query("predicted == 1")
    [["comment", "actual", "predicted", "toxic_probability"]]
    .head(10)
)

# Inspect false negatives with lowest confidence
print("\nFALSE NEGATIVES — LOW CONFIDENCE")

print(
    results[results["actual"] == 1]
    .sort_values("toxic_probability")
    .query("predicted == 0")
    [["comment", "actual", "predicted", "toxic_probability"]]
    .head(10)
)
```
