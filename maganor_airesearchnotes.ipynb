## 4. Baseline Model Development

### 4.1 Research Methodology and Experimental Design

The first stage of Maganor AI involved developing a traditional machine-learning model to establish a baseline for harmful language detection. This baseline provides a point of comparison for evaluating the performance of a more advanced transformer-based model, DistilBERT.

The experiment was designed as a binary text-classification task. The model receives an online comment as input and predicts whether it belongs to the toxic or non-toxic category.

The experiment followed these main stages:

1. **Data preparation:** The original comments and their toxicity labels were selected from the Jigsaw dataset.
2. **Data splitting:** The dataset was divided into training and testing subsets to evaluate the model on comments it had not previously encountered.
3. **Text representation:** TF-IDF was used to convert written comments into numerical features that a machine-learning algorithm could process.
4. **Model training:** Logistic Regression was trained using the extracted text features.
5. **Model evaluation:** Accuracy, precision, recall, F1-score, and a confusion matrix were used to assess performance.
6. **Error analysis:** False positives and false negatives were examined to understand situations in which the model made incorrect predictions.

This approach establishes a measurable baseline and provides an opportunity to investigate the limitations of traditional text-classification methods before introducing a transformer-based model.

### 4.2 Defining the Classification Target

The Jigsaw dataset contains six toxicity-related labels: toxic, severe toxic, obscene, threat, insult, and identity hate.

For the initial experiment, the `toxic` column was selected as the target variable. This decision simplified the task into binary classification, allowing the model to distinguish between comments labeled toxic and comments not labeled toxic.

The target variable was defined as follows:

- **0:** Non-toxic comment according to the dataset's `toxic` label.
- **1:** Toxic comment according to the dataset's `toxic` label.

It is important to clarify that this experiment does not independently identify every possible form of online harassment. It learns to predict the dataset's toxic label. The other five labels were retained in the dataset for further analysis.

### 4.3 Training and Testing Data

The dataset was divided into two subsets:

- **Training set:** 80% of the data, containing 127,656 comments.
- **Testing set:** 20% of the data, containing 31,915 comments.

A stratified train-test split was used to preserve approximately the same proportion of toxic and non-toxic comments in both subsets.

The experiment used a random state of 42 to make the split reproducible. This means that the same observations can be assigned to the training and testing sets when the experiment is repeated using the same settings.

The testing set was kept separate during model training so that it could provide an independent evaluation of the model's predictions.

### 4.4 TF-IDF Feature Extraction

Machine-learning algorithms generally require numerical input rather than raw text. Therefore, the comments were converted into numerical representations using Term Frequency–Inverse Document Frequency (TF-IDF).

TF-IDF measures how important a word or phrase is within a document relative to the collection of documents.

It considers two main factors:

- **Term Frequency (TF):** How frequently a word or phrase appears in a particular comment.
- **Inverse Document Frequency (IDF):** How informative a word or phrase is across the entire collection of comments. Words appearing in many comments generally receive lower importance.

For this experiment, the `TfidfVectorizer` from Scikit-learn was configured with the following parameters:

| Parameter | Setting | Purpose |
|---|---:|---|
| `max_features` | 50,000 | Limits the vocabulary to 50,000 features. |
| `ngram_range` | (1, 2) | Considers individual words and two-word combinations. |
| `min_df` | 2 | Ignores terms appearing in fewer than two documents. |
| `max_df` | 0.95 | Ignores terms appearing in more than 95% of documents. |
| `sublinear_tf` | True | Applies logarithmic scaling to term frequency. |

The vectorizer was fitted on the training comments only. The same fitted vectorizer was then used to transform the testing comments. This prevents information from the test set from influencing vocabulary construction.

The resulting numerical representations had the following dimensions:

- Training data: 127,656 comments × 50,000 features.
- Testing data: 31,915 comments × 50,000 features.

Using both individual words and two-word combinations allows the model to consider some short expressions rather than treating every word independently. However, TF-IDF does not fully understand sentence meaning, context, sarcasm, or the relationships between words.

### 4.6 Logistic Regression Model Training

After converting the comments into numerical features using TF-IDF, Logistic Regression was selected as the baseline classification algorithm.

Despite its name, Logistic Regression is commonly used for classification rather than predicting continuous numerical values.

In this experiment, the model learned patterns associated with the toxic and non-toxic categories from the training data. It then used these learned patterns to estimate the probability that an unseen comment belonged to the toxic category.

The model was configured with three important settings:

- **max_iter=1000:** Allows the optimization algorithm up to 1,000 iterations to find suitable model parameters.
- **class_weight="balanced":** Adjusts the importance of each class based on its frequency in the training data. This helps address the imbalance between toxic and non-toxic comments.
- **random_state=42:** Supports reproducibility.

Class balancing was particularly important because toxic comments represented only approximately 9.58% of the dataset, while non-toxic comments represented approximately 90.42%.

Without considering this imbalance, a model could achieve seemingly high accuracy by favoring the majority class while failing to identify many toxic comments.

The trained Logistic Regression model served as the first benchmark for evaluating Maganor AI's ability to classify harmful language.

### 4.8 Baseline Model Evaluation

The trained model was evaluated using the testing dataset, which contained 31,915 comments that were not used during training.

Five evaluation measures were considered:

1. **Accuracy:** The proportion of all predictions that were correct.
2. **Precision:** The proportion of comments predicted as toxic that were actually labeled toxic.
3. **Recall:** The proportion of dataset-labeled toxic comments that the model successfully identified.
4. **F1-score:** The harmonic mean of precision and recall, providing a balance between the two measures.
5. **Confusion matrix:** A summary of correct and incorrect predictions across both classes.

The model achieved the following results:

| Evaluation metric | Result |
|---|---:|
| Accuracy | 94.04% |
| Precision | 64.07% |
| Recall | 86.04% |
| F1-score | 73.45% |

The model achieved relatively high recall, meaning it identified a large proportion of comments labeled toxic in the testing dataset. However, its precision was lower, indicating that some comments predicted as toxic were not labeled toxic in the dataset.

These results demonstrate why accuracy alone is insufficient for evaluating a classifier trained on an imbalanced dataset.

Despite its useful performance, the TF-IDF and Logistic Regression model has several limitations.

# 5. DistilBERT Model Development

### 5.1 Introduction to Transformer-Based Models

Traditional machine-learning approaches, such as TF-IDF combined with Logistic Regression, represent text using statistical relationships between words and documents. Although this approach can identify useful vocabulary patterns, it has limitations when interpreting context and relationships between words.

Transformer-based language models offer a different approach. They use attention mechanisms to learn relationships between words within a sequence, allowing them to develop contextual representations of language.

For example, the meaning of a word may change depending on the surrounding sentence. A contextual language model can use information from the surrounding words to produce a more meaningful representation.

In this project, DistilBERT was selected as the second model to investigate whether contextual language representations could improve the classification of toxic comments.

The experiment will compare its performance against the TF-IDF and Logistic Regression baseline using the same testing dataset.

### 5.2 Why DistilBERT?

DistilBERT is a smaller, more computationally efficient version of BERT (Bidirectional Encoder Representations from Transformers).

It was developed using knowledge distillation, a process in which a smaller model learns to reproduce important behaviors of a larger model.

DistilBERT was selected for this project for several reasons:

1. **Contextual understanding:** It creates contextual representations rather than relying only on individual word frequencies.
2. **Computational efficiency:** It is smaller and generally faster than the original BERT model.
3. **Pretrained language knowledge:** It starts with knowledge learned from large-scale language pretraining, reducing the need to train a language model from scratch.
4. **Practical experimentation:** It provides an opportunity to investigate transformer-based classification within limited computing resources.

The project will use the pretrained `distilbert-base-uncased` model and fine-tune it on the Jigsaw toxicity dataset.

However, DistilBERT is not guaranteed to understand every form of sarcasm, indirect hostility, or culturally specific language. Its actual strengths and limitations will be examined through evaluation and error analysis.

```python
### 5.5 DistilBERT Tokenization

Tokenization is the process of converting written text into smaller units called tokens so that a language model can process it.

Unlike the TF-IDF approach, which represents text using word and phrase importance, DistilBERT uses a tokenizer designed for its pretrained vocabulary.

DistilBERT uses WordPiece tokenization. This means that a word can be represented as one token or divided into smaller subword units.

For example, a familiar word may be represented as a single token, while an unfamiliar or complex word may be divided into multiple pieces.

After tokenization, the text is converted into numerical token IDs. These IDs are then used by the model to generate contextual representations.

Tokenization also introduces special tokens and attention masks, which help the model identify the structure of its input and distinguish actual text from padding.
```

### 5.5.5 Maximum Sequence Length

Transformer models process input sequences using a predefined maximum length. For this experiment, the maximum sequence length was set to 128 tokens.

Comments longer than this limit will be truncated, while shorter comments will be padded when necessary to create consistent input dimensions.

This setting was selected as a practical starting point to balance computational efficiency and the amount of text available to the model.

An important limitation is that information appearing after the truncation point may be lost. Therefore, some context in longer comments may not be considered during classification.

### 5.9 Configuring the DistilBERT Training Process

Fine-tuning involves updating a pretrained model's parameters using examples from a specific task. In this experiment, DistilBERT will be fine-tuned to classify online comments as toxic or non-toxic.

Several training parameters must be selected before the process begins.

**Learning rate:** Determines how much the model's parameters are adjusted during each optimization step. A learning rate that is too high may cause unstable training, while one that is too low may result in slow learning.

**Batch size:** Determines how many examples the model processes before updating its parameters. A smaller batch size reduces memory requirements.

**Number of epochs:** Represents how many times the model passes through the entire training dataset.

**Optimizer:** The algorithm responsible for updating model parameters to reduce the training loss.

For the initial experiment, a relatively small training configuration will be used to balance computational requirements and model performance.

### 5.9.6 Handling Class Imbalance During Fine-Tuning

The Jigsaw dataset contains substantially more non-toxic comments than toxic comments. This imbalance is important because a model may become biased toward predicting the majority class.

### 5.10 Fine-Tuning DistilBERT

Fine-tuning is the process of adapting a pretrained language model to perform a specific task using a labeled dataset.

In this experiment, DistilBERT begins with language representations learned during its original pretraining. It is then trained using the Jigsaw dataset to identify patterns associated with toxic and non-toxic comments.

During training, the model processes batches of comments, generates predictions, calculates the difference between its predictions and the actual labels, and updates its parameters to reduce classification error.

The objective is to adapt the pretrained model to binary toxicity classification while preserving useful language representations learned during pretraining.

The initial experiment uses 20,000 training comments, a batch size of 8, and one training epoch.

### 5.10.4 Fine-Tuning Results

DistilBERT was fine-tuned using a sample of 20,000 labeled comments from the Jigsaw training dataset.

The experiment used a batch size of 8, a learning rate of 2e-5, and one training epoch.

The training process updated the model parameters to adapt its pretrained language representations to binary toxicity classification.

The training loss and execution time were recorded to document the experiment's computational performance.

These training results provide information about the learning process, while the model's actual classification performance will be assessed separately using the held-out testing dataset.

### 5.11 DistilBERT Model Evaluation

After fine-tuning, the DistilBERT model was evaluated using the held-out test dataset. This dataset contains comments that were not included in the training process.

The purpose of this evaluation is to measure how effectively the fine-tuned model identifies toxic and non-toxic comments on unseen data.

The model generates prediction scores (logits) for each test comment. These scores are converted into predicted class labels, which are then compared with the original labels to assess model performance.

### 5.11 DistilBERT Model Evaluation

After fine-tuning the pretrained DistilBERT model on 20,000 training comments, the model was evaluated using the same held-out test set of 31,915 comments used in the baseline experiment.

The model achieved an accuracy of 96.68%, precision of 84.95%, recall of 79.37%, and an F1-score of 82.07%.

Compared with TF-IDF and Logistic Regression, DistilBERT improved accuracy, precision, and F1-score. Its precision increased from 64.07% to 84.95%, indicating that its positive predictions were substantially more reliable. However, its recall decreased from 86.04% to 79.37%, meaning that it missed a greater proportion of comments labelled toxic.

These findings suggest that the pretrained language model provided a stronger overall balance between precision and recall in this initial experiment. Nevertheless, the comparison has an important limitation: Logistic Regression was trained on 127,656 examples, whereas DistilBERT was trained on only 20,000 examples. Therefore, the results should not be interpreted as a fully controlled comparison.

Further investigation of false positives and false negatives is necessary to understand the models' different classification behaviours.

### 5.12 Confusion Matrix Analysis

The DistilBERT model produced 28,426 true negatives, 430 false positives, 631 false negatives, and 2,428 true positives across the 31,915 test comments.

The results reveal an important difference between the two classification approaches. DistilBERT generated substantially fewer false positives than the TF-IDF and Logistic Regression baseline, reducing them from 1,476 to 430. However, its false negatives increased from 427 to 631.

This indicates that DistilBERT was more conservative when assigning the toxic label. Although this improved the reliability of its positive predictions, it also caused the model to overlook more comments that were labelled toxic in the dataset.

These findings demonstrate that high overall accuracy does not necessarily mean that a model identifies every harmful comment. Both false positives and false negatives matter, particularly in automated content moderation, where incorrect decisions can affect users and online discussions.

### 5.13 Qualitative Analysis of DistilBERT False Negatives

To investigate the limitations of DistilBERT, five examples were selected from the 631 false-negative predictions. These were comments labelled as toxic in the dataset but classified as non-toxic by the model.

The qualitative examination revealed several potentially challenging language patterns.

First, some comments contained multiple ideas, including personal attacks, accusations, and ordinary discussion. This suggests that identifying harmful language within longer passages may be challenging.

Second, certain comments used indirect insults, sarcasm, or expressions whose meanings depended on the surrounding conversation. Such examples demonstrate that harmful meaning cannot always be determined through individual words.

Third, some comments mixed insulting expressions with joking language, making their intended tone ambiguous. Another example contained antisemitic language within a historical quotation, highlighting the importance of distinguishing quoted harmful content from the context and purpose of the surrounding discussion.

These observations suggest that automated toxicity classification involves more than identifying offensive vocabulary. Understanding context, tone, indirect expressions, and quoted material remains challenging.

However, this analysis is exploratory and based on a small selection of false negatives. Further examination of a larger sample and human review would be necessary before drawing broader conclusions.

### 5.14 Confidence Analysis of False-Negative Predictions

To investigate the confidence of DistilBERT's incorrect predictions, the predicted probability of the toxic class was examined for all 631 false-negative comments.

The average toxic probability was 0.0787 (7.87%), while the median was 0.0221 (2.21%). The probabilities ranged from 0.000543 to 0.492833.

These findings indicate that many false-negative predictions were made with high model confidence rather than occurring only near the classification threshold.

The examination of the most confident mistakes revealed that some comments contained direct insults, aggressive expressions, and extended hostile discussions. Nevertheless, DistilBERT assigned them very low probabilities of belonging to the toxic class.

This observation raises questions about how pretrained language models learn associations between words, context, and toxicity labels. It also demonstrates that high confidence does not necessarily guarantee a correct classification.

However, the probability analysis alone cannot establish why these errors occurred. Further experiments involving contextual examples, model explanations, and human review would be necessary to investigate the underlying causes.

### 5.15 Investigation of Comment Length and Classification Errors

The relationship between comment length and DistilBERT classification errors was investigated by comparing false-negative comments with correctly identified toxic comments.

The 631 false-negative comments had an average length of 408.31 characters and a median length of 167 characters. In comparison, the 2,428 correctly identified toxic comments had an average length of 268.55 characters and a median length of 111 characters.

The false-negative group therefore contained longer comments on average. This observation suggests a possible association between comment length and classification difficulty.

One potential explanation is input truncation. The DistilBERT tokenizer was configured with a maximum sequence length of 128 tokens. Consequently, longer comments could lose information beyond the model's input limit. If important expressions appeared in the truncated portion, the model would not be able to use them during classification.

Another possible explanation is that longer comments contain a mixture of ordinary discussion and harmful expressions, making their overall meaning more difficult to interpret.

These findings do not establish that comment length caused the classification errors. Further experiments using different sequence lengths would be required to investigate the potential influence of truncation.

### 5.16 Relationship Between Input Length and False-Negative Rates

To investigate whether input length was associated with classification errors, all 3,059 toxic comments in the test set were divided into two groups according to the tokenizer's maximum sequence length of 128 tokens.

Among the 2,675 comments containing 128 tokens or fewer, 500 were incorrectly classified as non-toxic, resulting in a false-negative rate of 18.69%.

Among the 384 comments exceeding 128 tokens, 131 were incorrectly classified as non-toxic, resulting in a false-negative rate of 34.11%.

The false-negative rate for longer comments was approximately 1.83 times that of shorter comments, representing a difference of 15.42 percentage points.

These findings indicate an association between longer input sequences and a higher rate of missed toxic comments in this experiment. One possible contributing factor is tokenizer truncation, which may remove relevant information from longer comments.

However, this analysis does not establish truncation as the direct cause of the increased error rate. Other linguistic and contextual factors may also influence classification performance. A controlled experiment using a longer maximum sequence length would help investigate this possibility.

### 5.17 Statistical Analysis of Comment Length and Classification Errors

A chi-square test of independence was conducted to investigate the relationship between comment length and DistilBERT classification outcomes.

The test produced a chi-square statistic of 47.8484, with one degree of freedom and a p-value of approximately 4.60 × 10⁻¹².

Since the p-value was substantially below the significance threshold of 0.05, the null hypothesis of independence was rejected.

The results provide strong statistical evidence of an association between comment length group and classification outcome. Toxic comments exceeding 128 tokens had a false-negative rate of 34.11%, compared with 18.69% for comments containing 128 tokens or fewer.

This finding supports further investigation into the effect of input sequence length on toxicity classification. However, statistical association does not establish causation. In particular, the experiment does not prove that tokenizer truncation alone caused the higher error rate.

A controlled experiment using a longer input sequence length is therefore proposed to investigate whether retaining more textual information can improve classification performance.

## Chapter 6: Discussion of Findings

### 6.1 Comparison of the Three Classification Models

The experiments demonstrated differences in how traditional machine learning and transformer-based models classified toxic online comments.

The TF-IDF and Logistic Regression baseline achieved an accuracy of 94.04%, precision of 64.07%, recall of 86.04%, and F1-score of 73.45%.

The first DistilBERT experiment, using a maximum input length of 128 tokens, achieved an accuracy of 96.68%, precision of 84.95%, recall of 79.37%, and F1-score of 82.07%.

The second DistilBERT experiment, using 256 tokens, achieved an accuracy of 96.64%, precision of 84.76%, recall of 79.24%, and F1-score of 81.91%.

These results indicate that DistilBERT achieved higher precision and F1-score than the baseline in the experiments conducted. However, Logistic Regression achieved higher recall, identifying a larger proportion of the toxic comments.

This difference illustrates that model performance depends on the evaluation metric being considered. A model that produces fewer false alarms does not necessarily identify more of the harmful comments in a dataset.

An important methodological consideration is that the baseline model was trained on 127,656 examples, whereas each DistilBERT experiment used 20,000 training examples. Consequently, the results represent the performance of the specific experimental configurations rather than a perfectly controlled comparison of model architectures.

### 6.2 The Effect of Increasing Input Length

One objective of the second DistilBERT experiment was to investigate whether increasing the maximum input length could improve toxic comment classification.

The results did not demonstrate an improvement. The F1-score decreased slightly from 82.07% at 128 tokens to 81.91% at 256 tokens.

The difference was small, and the results should not be interpreted as proof that shorter inputs are generally better. Nevertheless, the experiment demonstrated that providing a model with more text does not automatically guarantee improved classification performance.

This observation raises further questions about how language models process context and whether additional information is always useful for identifying harmful expressions.

### 6.3 Comment Length and False-Negative Predictions

The error analysis revealed an association between comment length and false-negative predictions in both DistilBERT experiments.

In the 128-token experiment, comments exceeding 128 tokens had a false-negative rate of 34.11%, compared with 18.69% for shorter comments.

In the 256-token experiment, comments exceeding 256 tokens had a false-negative rate of 34.03%, compared with 20.10% for shorter comments.

Chi-square tests indicated statistically significant associations in both experiments. However, the corresponding Cramér's V values (0.125 and 0.071) indicated small associations.

These findings suggest that longer comments remain challenging for the models. Truncation may contribute to this difficulty because portions of longer comments are not processed, but other factors, including linguistic complexity and contextual ambiguity, may also influence classification errors.

The findings establish an association rather than a causal relationship.

### 6.4 Understanding False Positives and False Negatives

The qualitative error analysis revealed two important limitations of automated toxic language classification.

False positives occurred when comments were predicted as toxic despite having a non-toxic primary label in the dataset. Some of these comments contained potentially offensive expressions or secondary toxicity labels, demonstrating that annotation decisions and the meaning of individual words can be complex.

False negatives occurred when comments labelled as toxic were classified as non-toxic. Some examples contained personal attacks, indirect insults, sarcastic expressions, or hostile language embedded in longer discussions.

These observations suggest that identifying harmful language is not simply a matter of detecting individual offensive words. The surrounding discussion, intended meaning, and linguistic context can influence how an expression should be interpreted.

However, individual examples cannot establish that a model consistently fails on a particular linguistic category.

### 6.5 Implications for Automated Content Moderation

The experiments demonstrate both the potential and limitations of automated language classification.

DistilBERT achieved stronger precision and F1-score than the TF-IDF baseline, but it still failed to identify some comments labelled as toxic. Meanwhile, the baseline achieved higher recall but generated more false-positive predictions.

In a practical moderation environment, these trade-offs matter. False positives can incorrectly flag legitimate discussions, while false negatives can allow harmful content to remain undetected.

Therefore, automated classification should be treated as a decision-support mechanism rather than a complete replacement for human judgment.

Human review remains important, particularly when comments contain sarcasm, indirect hostility, ambiguous expressions, or culturally specific language.

## Chapter 7: Research Limitations and Responsible AI Considerations

### 7.1 Dataset Limitations

This research used the Jigsaw Toxic Comment Classification dataset, which contains English-language comments from Wikipedia discussions.

Although the dataset provides useful examples for studying toxic language classification, it does not represent all forms of online communication.

In particular, the dataset does not adequately represent Ethiopian social-media environments, local cultural expressions, or languages such as Amharic and Afaan Oromo.

Consequently, the models developed in this project cannot be assumed to identify harmful language accurately across Ethiopian online communities.

The dataset also contains human-assigned labels, which may not capture every interpretation of a comment. Some expressions may be ambiguous, context-dependent, or interpreted differently by different annotators.

### 7.2 Model and Experimental Limitations

The experiments were conducted under limited computational and time resources.

The Logistic Regression model used 127,656 training examples, while each DistilBERT experiment used a randomly selected subset of 20,000 examples. This difference limits the fairness of direct architectural comparisons.

Additionally, each DistilBERT configuration was trained for only one epoch. The experiments were not repeated across multiple random seeds, and no hyperparameter search was conducted.

Therefore, the observed differences between the 128-token and 256-token experiments should be interpreted as findings from individual experimental runs rather than definitive evidence that one sequence length is universally better.

The evaluation focused on binary toxic versus non-toxic classification using the primary toxic label. Although the dataset contains six toxicity-related labels, the project did not develop a complete multi-label classification system.

### 7.3 Limitations of Contextual Understanding

The error analysis identified false-positive and false-negative predictions involving potentially ambiguous or context-dependent language.

However, the project did not independently establish the intended meaning of every comment or conduct a formal human annotation study.

The observed errors therefore provide examples of classification limitations, not proof that the models cannot understand sarcasm, indirect hostility, or gender-based harassment in every situation.

Further investigation would require carefully selected examples, additional human evaluation, and datasets representing different linguistic and cultural contexts.

### 7.4 Ethical Considerations

Automated toxicity classification can influence how people communicate and participate in online spaces.

Incorrectly classifying legitimate comments as harmful may discourage participation, while failing to identify genuinely harmful expressions may leave affected individuals without adequate protection.

There is also a risk of bias when a model is trained on data that does not adequately represent the communities in which it may eventually be used.

For these reasons, the results of this project should not be used as the sole basis for deleting content, restricting accounts, or making judgments about individuals.

Any future practical application should include appropriate human oversight, transparent explanations of system limitations, and opportunities to challenge incorrect classifications.

### 7.5 Privacy and Responsible Data Use

The research uses an existing public research dataset rather than collecting private conversations or personal messages.

The project focuses on classifying textual content, not identifying individuals or inferring their personal characteristics.

Any future extension involving social-media data collection should consider informed consent where appropriate, data minimization, secure storage, and protection against exposing sensitive personal information.

### 7.6 Future Research Directions

Several areas could strengthen this research in the future:

- Investigating how sarcasm, indirect hostility, and context influence toxic language classification.
- Evaluating models on carefully curated Ethiopian social-media examples, with appropriate privacy safeguards.
- Exploring multilingual toxicity classification for Amharic and Afaan Oromo.
- Conducting human evaluation to examine ambiguous predictions and annotation disagreements.
- Testing model performance across different training configurations and random seeds.
- Investigating methods for improving the detection of long and context-dependent comments.

These directions represent opportunities for further research rather than capabilities already implemented in the current prototype.
